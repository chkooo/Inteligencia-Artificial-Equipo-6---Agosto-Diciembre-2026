En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

*   `2022 Ranking`: Ranking del año actual de la empresa

*   `Assets (Billions)`: Activos del año

### **Gráficas de Barras para Columnas Categóricas**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar plot for Industry
plt.figure(figsize=(12, 6))
sns.barplot(x=forbes_df['Industry'].value_counts().index, y=forbes_df['Industry'].value_counts().values)
plt.xticks(rotation=90)
plt.title('Frecuencia de Industrias')
plt.xlabel('Industria')
plt.ylabel('Cantidad de Empresas')
plt.show()
```

```python
# Bar plot for Country (top 10 countries)
plt.figure(figsize=(12, 6))
sns.barplot(x=forbes_df['Country'].value_counts().head(10).index, y=forbes_df['Country'].value_counts().head(10).values)
plt.xticks(rotation=45, ha='right')
plt.title('Top 10 Países con Más Empresas')
plt.xlabel('País')
plt.ylabel('Cantidad de Empresas')
plt.show()
```

### **Gráficas Boxplot para Valores Numéricos**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Box plots for numerical columns
numerical_cols = ['Year Founded', 'Revenue (Billions)', 'Profits (Billions)', 'Assets (Billions)', 'Market Value (Billions)', 'Total Employees']

for col in numerical_cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(y=forbes_df[col])
    plt.title(f'Boxplot de {col}')
    plt.ylabel(col)
    plt.show()
```

### **Gráficas de Dispersión**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Scatter plot for Assets (Billions) vs Market Value (Billions)
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Assets (Billions)', y='Market Value (Billions)', data=forbes_df)
plt.title('Relación entre Activos y Valor de Mercado')
plt.xlabel('Activos (Billones)')
plt.ylabel('Valor de Mercado (Billones)')
plt.show()
```

### **Mapa de Calor de Correlación**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Heatmap of correlations between numerical columns
plt.figure(figsize=(10, 8))
sns.heatmap(forbes_df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Mapa de Calor de Correlación entre Columnas Numéricas')
plt.show()
```
