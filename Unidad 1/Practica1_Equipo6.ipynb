En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

*   `2022 Ranking`: Ranking del año actual de la empresa
*   `Organization Name`: Nombre
*   `Industry`: El tipo de industria con la que trata principalmente
*   `Country`: País de origen
*   `Year Founded`: Año en que se fundó
*   `CEO`: CEO
*   `Profits (Billions)`: Ganancias del año
*   `Assets (Billions)`: Activos del año
*   `Market Value (Billions)`: Valor de mercado del año
*   `Total Employees`: Total de empleados

https://www.kaggle.com/datasets/ash316/forbes-top-2000-companies

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Figura con una cuadrícula de 3x2
fig, axes = plt.subplots(3, 2, figsize=(18, 16))

fig.delaxes(axes[2][1])

plot_df = forbes_df.reset_index()

# Top 10 Industrias
top_industries = plot_df['Industry'].value_counts().nlargest(10).index
sns.countplot(y='Industry', data=plot_df, order=top_industries, ax=axes[0,0], palette='viridis', hue='Industry', legend=False)
axes[0,0].set_title('Top 10 Industrias con más empresas')
axes[0,0].set_xlabel('Cantidad de Empresas')
axes[0,0].set_ylabel('Industria')

# Top 10 Países
top_countries = plot_df['Country'].value_counts().nlargest(10).index
sns.countplot(y='Country', data=plot_df, order=top_countries, ax=axes[0,1], palette='magma', hue='Country', legend=False)
axes[0,1].set_title('Top 10 Países con más empresas')
axes[0,1].set_xlabel('Cantidad de Empresas')
axes[0,1].set_ylabel('País')

# Top 5 Empresas por Ingresos
top5_revenue = plot_df.nlargest(5, 'Revenue (Billions)')
sns.barplot(x='Revenue (Billions)', y='Organization Name', data=top5_revenue, ax=axes[1,0], palette='flare', hue='Organization Name', legend=False)
axes[1,0].set_title('Top 5 Empresas por Ingresos (Billions)')
axes[1,0].set_ylabel('')

# Top 5 Empresas por Ganancias
top5_profits = plot_df.nlargest(5, 'Profits (Billions)')
sns.barplot(x='Profits (Billions)', y='Organization Name', data=top5_profits, ax=axes[1,1], palette='crest', hue='Organization Name', legend=False)
axes[1,1].set_title('Top 5 Empresas por Ganancias (Billions)')
axes[1,1].set_ylabel('')

# Top 5 Empresas por Total de Empleados
top5_employees = plot_df.nlargest(5, 'Total Employees')
sns.barplot(x='Total Employees', y='Organization Name', data=top5_employees, ax=axes[2,0], palette='cubehelix', hue='Organization Name', legend=False)
axes[2,0].set_title('Top 5 Empresas por Total de Empleados')
axes[2,0].set_ylabel('')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = forbes_df.select_dtypes(include=['number']).columns

# Calcular cantidad de filas y columnas
n_cols = 2
n_rows = (len(num_cols) + 1) // n_cols

# Figura estructurada como dashboard
fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4 * n_rows))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    sns.boxplot(data=forbes_df, x=col, ax=axes[i], color='skyblue')
    axes[i].set_title(f'Distribución de {col}')
    axes[i].set_xlabel(col)

for j in range(len(num_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes = axes.flatten()

# Ingresos vs Ganancias
sns.scatterplot(data=forbes_df, x='Revenue (Billions)', y='Profits (Billions)',
                ax=axes[0], color='darkblue', alpha=0.6)
axes[0].set_title('Ingresos vs Ganancias')
axes[0].set_xlabel('Ingresos (Billones)')
axes[0].set_ylabel('Ganancias (Billones)')

# Activos vs Valor de Mercado
sns.scatterplot(data=forbes_df, x='Assets (Billions)', y='Market Value (Billions)',
                ax=axes[1], color='darkgreen', alpha=0.6)
axes[1].set_title('Activos vs Valor de Mercado')
axes[1].set_xlabel('Activos (Billones)')
axes[1].set_ylabel('Valor de Mercado (Billones)')

# Ingresos vs Valor de Mercado
sns.scatterplot(data=forbes_df, x='Revenue (Billions)', y='Market Value (Billions)',
                ax=axes[2], color='purple', alpha=0.6)
axes[2].set_title('Ingresos vs Valor de Mercado')
axes[2].set_xlabel('Ingresos (Billones)')
axes[2].set_ylabel('Valor de Mercado (Billones)')

# Ganancias vs Valor de Mercado
sns.scatterplot(data=forbes_df, x='Profits (Billions)', y='Market Value (Billions)',
                ax=axes[3], color='darkred', alpha=0.6)
axes[3].set_title('Ganancias vs Valor de Mercado')
axes[3].set_xlabel('Ganancias (Billones)')
axes[3].set_ylabel('Valor de Mercado (Billones)')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

corr_matrix = forbes_df.select_dtypes(include=['number']).corr()

plt.figure(figsize=(10, 8))

# Mapa de calor con límites de correlación fijos
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)

plt.title('Mapa de Calor de Correlaciones Numéricas')
plt.tight_layout()
plt.show()
```
